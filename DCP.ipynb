```python
import pandas as pd

df = pd.read_csv("gen_data.csv")

print(df.head())
print(df.info())
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
```

```python
num_cols = df.select_dtypes(include="number").columns
catg_cols = df.select_dtypes(include=["object", "category"]).columns

print((num_cols))
print((catg_cols))
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder



si_num = SimpleImputer(strategy="median")
df[num_cols] = si_num.fit_transform(df[num_cols])


si_cat = SimpleImputer(strategy="most_frequent")


'''print("Missing values after cleaning")
print(df.isnull().sum())
'''
```

```python
from sklearn.pipeline import Pipeline

num_pipeline = Pipeline([
    ("imputer", si_num)
])
cat_pipeline = Pipeline([
    ("imputer", si_cat)])
#------------------------------------------------------------
df[num_cols] = num_pipeline.fit_transform(df[num_cols])
df[catg_cols] = cat_pipeline.fit_transform(df[catg_cols])

#------------------------------------------------------------------
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
#en cat data
df = pd.get_dummies(df, columns=catg_cols, dtype=int)

''' too create a new file with clean data '''
df.to_csv("clean_data.csv", index=False)

print("done")
```
